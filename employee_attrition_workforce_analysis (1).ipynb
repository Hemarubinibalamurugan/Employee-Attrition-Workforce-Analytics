#Attrition by Department

```python
sns.countplot(data=df, x='Department', hue='Attrition')
plt.show()
```

Interpretation: Attrition by Department

- Chart: Compares the number of employees who stayed (No) and left (Yes) across six departments.
- Key Findings: Research & Development shows a higher count of employees who left than stayed. Hardware shows a slightly higher count of employees who stayed, while Human Resources, Sales, Software, and Support show relatively small differences between the two groups.
- Business Insight: The visible differences suggest that employee attrition patterns vary across departments. Research & Development may require closer investigation to understand the factors contributing to employee turnover.
- Recommendation: HR should compare department-wise attrition rates and investigate factors such as job satisfaction, overtime, workload, and salary to develop targeted employee retention strategies.

```python
pd.crosstab(df['Department'], df['Attrition'])
```

Interpretation: Attrition by Department

- Chart: Compares employees who stayed (No) and left (Yes) across six departments.
- Key Findings: Research & Development recorded 4,260 employees who left compared with 4,059 who stayed, showing the largest visible difference in favour of attrition. Support recorded 4,168 employees who left and 4,137 who stayed. Hardware had 4,130 employees who stayed compared with 4,039 who left. Human Resources and Sales showed slightly higher counts of employees who stayed, while Software had equal counts of 4,123 in both categories.
- Business Insight: Research & Development shows the largest difference between the two groups, so it may warrant further investigation. However, these are counts rather than attrition rates, so they do not establish which department has the highest likelihood of employee turnover.
- Recommendation: HR should calculate department-wise attrition percentages and investigate factors such as job satisfaction, workload, overtime, and salary to identify potential causes of employee turnover.

Final Insights and Conclusion

Final Insights

- The analysis indicates an approximately 50% attrition rate in the displayed dataset, suggesting a nearly equal split between employees who left and those who stayed. This unusually high rate should be validated against the original data before drawing business conclusions.

- Employee attrition counts are relatively similar across departments and job roles, indicating that turnover is not visibly concentrated in a single department or role within the displayed results.

- Attrition counts vary across age groups, with some younger and middle-age groups showing higher employee counts than older groups. Further comparison of group-level attrition rates would be needed to determine which age groups face greater turnover risk.

- Employees who left the organization show a slightly higher median monthly income but a slightly lower median tenure than employees who stayed. These patterns suggest that compensation alone may not explain employee turnover and that employment experience may also warrant investigation.

- Overtime, job satisfaction, work-life balance, business travel, and distance from home show broadly similar attrition and retention counts or distributions in the displayed visualizations. These results do not establish a strong relationship between any one of these factors and attrition.

- Overall, the findings suggest that employee turnover should be examined through a combination of demographic, job-related, compensation, and work-experience factors rather than attributing it to a single variable.

Conclusion

This project provides a structured analysis of employee attrition by examining workforce patterns across multiple employee characteristics. The visualizations highlight differences in tenure, income, age distribution, and workforce composition while showing broadly similar attrition counts across several other categories. However, the displayed attrition rate of approximately 50% is unusually high for a typical workforce dataset and requires validation. Once data quality is confirmed, further analysis of attrition rates by employee segment can help identify meaningful turnover patterns and support evidence-based workforce planning.
